```python
frequency = 3665e6
capture_statistic = "max"
detector = "rms"
pfp_ = day_block.get_aligned_pfp(channel=frequency,capture_statistic=capture_statistic,detector=detector)
max_pfp = max(pfp_.max())
x_axis = np.linspace(0,10,len(pfp_.columns))
text_y = -67.5
y_offset_plot = 2.5
noise_figure_estimate = np.median(day_block.noise_figures[frequency])
noise_floor_estimate = -174 + 70 + noise_figure_estimate
plt.fill_between(x_axis,np.median(pfp_,axis=0),noise_floor_estimate,label="Median",hatch='/',color="grey",alpha=.5)
#plt.plot(x_axis,np.mean(pfp_),label="Mean")
plt.axvline(2,color="r",linewidth=5,linestyle="dashed")
plt.axvline(3,color="r",linewidth=5,linestyle="dashed")

plt.axvline(7,color="r",linewidth=5,linestyle="dashed")
plt.axvline(8,color="r",linewidth=5,linestyle="dashed")
plt.axhline(noise_floor_estimate,color="k",linewidth=5,linestyle="dashed")
plt.text(2.25,text_y,"DL")
plt.text(7.25,text_y,"DL")
plt.text(.75,text_y,"UL")
plt.text(4.75,text_y,"UL")
plt.text(8.75,text_y,"UL")

plt.ylim([noise_floor_estimate-y_offset_plot,max_pfp+y_offset_plot])
plt.xlabel("Time (ms)")
plt.ylabel(r"$Power (\frac{dBm}{10 MHz})$")
#plt.legend(loc="lower right")
plt.title(f"{day_block.date} - {day_block.sensor}, {frequency*1e-6} MHz, median of {detector}-{capture_statistic}")
plt.grid()
```

```python
import matplotlib.gridspec

frequency = 3665e6
capture_statistic = "max"
detector = "rms"
bins = 50
pfp_ = day_block.get_aligned_pfp(channel=frequency,capture_statistic=capture_statistic,detector=detector)
median_pfp = np.median(pfp_,axis=0)
max_pfp = max(pfp_.max())
x_axis = np.linspace(0,10,len(pfp_.columns))
ul1_start = np.argmin(abs(x_axis-2.0))
ul1_stop = np.argmin(abs(x_axis-3.0))
ul2_start = np.argmin(abs(x_axis-7.0))
ul2_stop = np.argmin(abs(x_axis-8.0))
ul_mask = np.zeros(len(x_axis),dtype=bool)
for i in range(len(x_axis)):
     if ul1_start<i<ul1_stop or ul2_start<i<ul2_stop:
          ul_mask[i]=True
     
uplink = pfp_.iloc[:,ul_mask]
downlink= pfp_.iloc[:,~ul_mask]

text_y = median_pfp.max()+2
y_offset_plot = 2.5
noise_figure_estimate = np.median(day_block.noise_figures[frequency])
noise_floor_estimate = -174 + 70 + noise_figure_estimate
figure = plt.figure(figsize=(10,4))
gs = matplotlib.gridspec.GridSpec(2,3)
ax1 = figure.add_subplot(gs[:, 0:2])
ax2 = figure.add_subplot(gs[0, 2])
ax3 = figure.add_subplot(gs[1, 2],sharex=ax2)

ax1.fill_between(x_axis,median_pfp,noise_floor_estimate,label="Median",hatch='/',color="grey",alpha=.5)
ax1.fill_between(x_axis[ul_mask],median_pfp[ul_mask],noise_floor_estimate,
                 label="Median",hatch='/',color="green",alpha=.5,where =(x_axis[ul_mask]< 3) )
ax1.fill_between(x_axis[ul_mask],median_pfp[ul_mask],noise_floor_estimate,
                 label="Median",hatch='/',color="green",alpha=.5,where =(x_axis[ul_mask]> 4) )
                 

#plt.plot(x_axis,np.mean(pfp_),label="Mean")
ax1.axvline(2,color="r",linewidth=5,linestyle="dashed")
ax1.axvline(3,color="r",linewidth=5,linestyle="dashed")
ax1.axvline(7,color="r",linewidth=5,linestyle="dashed")
ax1.axvline(8,color="r",linewidth=5,linestyle="dashed")
ax1.axhline(noise_floor_estimate,color="k",linewidth=5,linestyle="dashed")
ax1.text(2.25,text_y,"UL")
ax1.text(7.25,text_y,"UL")
ax1.text(.75,text_y,"DL")
ax1.text(4.75,text_y,"DL")
ax1.text(8.75,text_y,"DL")
ax1.set_ylim([noise_floor_estimate-y_offset_plot,max_pfp+y_offset_plot])
ax1.set_xlabel("Time (ms)")
ax1.set_ylabel(r"$Power (\frac{dBm}{10 MHz})$")
ax1.set_title(f"{day_block.date} - {day_block.sensor}, {frequency*1e-6} MHz, median of {detector}-{capture_statistic}")
ax1.grid()
ax2.hist(uplink.values.flatten(),density=True,bins=bins,hatch='/',color="green",alpha=.5)
ax2.text(0.5, 0.5, 'UL', ha='center', va='center', transform=ax2.transAxes)
ax2.grid()
ax3.hist(downlink.values.flatten(),density=True,bins=bins,hatch='/',color="gray",alpha=.5)
ax3.text(0.5, 0.5, 'DL', ha='center', va='center', transform=ax3.transAxes)
ax3.grid()
ax3.set_xlabel(r"$Power (\frac{dBm}{10 MHz})$")

plt.tight_layout()
```
